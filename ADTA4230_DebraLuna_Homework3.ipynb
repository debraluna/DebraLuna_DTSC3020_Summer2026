```python
#Q2

from numpy import astype
import pandas as pd
import statsmodels.formula.api as smf
import statsmodels.api as sm
from statsmodels.api import families
from sklearn.metrics import accuracy_score

# Load the dataset
myData = pd.read_excel('Ch9_Q43_V18_Data_File(1).xlsx', sheet_name = '18')

myData.head()

# Split data
TData = myData[:300]
VData = myData[300:500]

# Training data
# The 'Default' column should remain in TData for the glm formula
# smf.glm usually handles the constant automatically or it can be added to the formula

VData_Default = VData['Default'] # Assign 'Default' column for validation separately

#fit logistic regression model
Model1 = smf.glm(formula = 'Default ~ LTV + FICO + Age', data = TData, family = families.Binomial()).fit()
Model1.summary()

# Validation data
X_valid = sm.add_constant(VData[['LTV', 'FICO', 'Age']], has_constant='add')

# Predicted probabilities
prob = Model1.predict(X_valid)

# Apply 0.50 cutoff
pred = (prob >= 0.50).astype(int)

# Validation accuracy
accuracy = accuracy_score(VData['Default'], pred)

print("Validation Accuracy =", accuracy)
```

```python
# Q4

from numpy import astype
import pandas as pd
import statsmodels.formula.api as smf
import statsmodels.api as sm
from statsmodels.api import families

myData = pd.read_excel('Ch9_Q23_Data_File.xlsx') # Removed sheet_name parameter to load the first sheet by default

myData.head()
myData['Trend'] = myData.index + 1
myData.head()

# encode categorical plan
myData["Single"] = (myData["Plan"] == "Single").astype(int)

# Fit logistic regression
x = myData[["Age", "Income", "Single"]]
x = sm.add_constant(x)
y = myData["Loyal"]

model = sm.Logit(y, x).fit()
print(model.summary())

# Predict and convert binary
new = pd.DataFrame({
    "const": [1],
    "Age": [50],
    "Income": [80],
    "Single": [1]
})

# Get probability
p = model.predict(new)[0]
odds = p / (1 - p)

print("Probability:", p)
print("Odds:", round(odds, 2))
```
